# Fraud transaction EDA
Synthetic data generated by src/generate.py. Run all cells from the project folder or notebooks folder.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
root = Path('..') if Path('../data/transactions.csv').exists() else Path('.')
df = pd.read_csv(root/'data/transactions.csv')
df.head()
```

```python
pd.DataFrame({'missing': df.isna().sum(), 'unique': df.nunique()})
```

```python
df.groupby('is_fraud').agg(transactions=('transaction_id','count'), mean_amount=('amount','mean'))
```

```python
df.assign(day=pd.to_datetime(df.timestamp).dt.date).groupby('day').is_fraud.agg(['count','sum','mean'])
```

```python
df.groupby('country').is_fraud.mean().plot.bar(title='Synthetic fraud rate by transaction country'); plt.ylabel('Fraud rate'); plt.show()
```

```python
df.boxplot(column='amount', by='is_fraud'); plt.yscale('log'); plt.show()
```

## Interpretation
Compare rates with group sample sizes. Synthetic country differences reflect the generator, not evidence about real populations. Model metrics are in reports/metrics.json. Never fit preprocessing to the entire dataset before splitting.
